## Implementación y Código Base

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

def ajuste_lineal(x, y):
    """
    Realiza ajuste lineal y retorna parámetros importantes
    """
    slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
    return {
        'pendiente': slope,
        'intercepto': intercept,
        'r_cuadrado': r_value**2,
        'error_std': std_err
    }

def graficar_ajuste(x, y, params, titulo="Ajuste Lineal"):
    """
    Grafica datos y línea de ajuste
    """
    plt.figure(figsize=(10, 6))
    plt.scatter(x, y, color='blue', label='Datos')
    
    x_line = np.linspace(min(x), max(x), 100)
    y_line = params['pendiente'] * x_line + params['intercepto']
    plt.plot(x_line, y_line, 'r-', label='Ajuste')
    
    plt.title(titulo)
    plt.legend()
    plt.grid(True)
    return plt

def analisis_residuos(x, y, params):
    """
    Calcula y grafica residuos
    """
    y_pred = params['pendiente'] * np.array(x) + params['intercepto']
    residuos = y - y_pred
    
    plt.figure(figsize=(10, 6))
    plt.scatter(x, residuos)
    plt.axhline(y=0, color='r', linestyle='--')
    plt.title('Análisis de Residuos')
    plt.grid(True)
    return residuos
```

## Solución Detallada: Ejercicio 1.1

```python
# Datos
temp = np.array([0, 20, 40, 60, 80, 100])
vol = np.array([2.0, 2.4, 2.8, 3.2, 3.6, 4.0])

# Ajuste
resultados = ajuste_lineal(temp, vol)

# Gráfica
plt = graficar_ajuste(temp, vol, resultados, "Volumen vs Temperatura")

# Predicción a 50°C
vol_50 = resultados['pendiente'] * 50 + resultados['intercepto']

print(f"""
Resultados del ajuste:
---------------------
Pendiente: {resultados['pendiente']:.4f} mL/°C
Intercepto: {resultados['intercepto']:.4f} mL
R²: {resultados['r_cuadrado']:.4f}
Error estándar: {resultados['error_std']:.4f}
Volumen estimado a 50°C: {vol_50:.2f} mL
""")
```
